```python
import os
import glob
import numpy as np
import tifffile
import napari
```

```python
# create a def to visualize the results for a given case, then loop over all cases in the validation set
def visualize_case(case, val_dir, images_dir, labels_dir):
    #load the predicted segmentation, the ground truth label, and the raw image
    pred  = tifffile.imread(os.path.join(val_dir,    f"{case}.tiff")).astype(np.int16)
    label = tifffile.imread(os.path.join(labels_dir, f"{case}.tiff")).astype(np.int16)
    raw   = tifffile.imread(os.path.join(images_dir, f"{case}_0000.tiff"))

    #check shapes match
    assert pred.shape == label.shape, f"shape mismatch: pred {pred.shape} vs label {label.shape}"

    diff = label - pred   # 0 = match, +1 = false negative, -1 = false positive
    
    viewer = napari.Viewer()
    viewer.add_image(raw,  name=f"{case}_raw", colormap="gray")
    viewer.add_image(diff, name=f"{case}_error", colormap="PiYG",
                     contrast_limits=(-1, 1), opacity=0.6)
    viewer.add_image(pred, name=f"{case}_pred", colormap="PiYG",
                     contrast_limits=(-1, 1), opacity=0.6)
    viewer.add_image(label, name=f"{case}_label", colormap="PiYG",
                     contrast_limits=(-1, 1), opacity=0.6)
    return viewer

```
